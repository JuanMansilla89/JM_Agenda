```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Rutas
DATA_PATH = Path('../Documentacion/Referencias/Datos Demo')
BH_PATH   = DATA_PATH / 'Data_BH'
DRILL_PATH = DATA_PATH / 'Data_drill'

print('DATA_PATH existe:', DATA_PATH.exists())
print('BH_PATH existe:  ', BH_PATH.exists())
print('DRILL_PATH existe:', DRILL_PATH.exists())
```

## 1. Cargar todos los archivos

```python
# --- BH ---
bh_collar  = pd.read_csv(BH_PATH / 'collar.csv')
bh_assay   = pd.read_csv(BH_PATH / 'assay.csv')
bh_litho   = pd.read_csv(BH_PATH / 'litho.csv')
bh_survey  = pd.read_csv(BH_PATH / 'survey.csv')

# --- Diamantina ---
dr_collar  = pd.read_csv(DRILL_PATH / 'collar.csv')
dr_assay   = pd.read_csv(DRILL_PATH / 'assay.csv')
dr_lith    = pd.read_csv(DRILL_PATH / 'lith.csv')
dr_drlrun  = pd.read_csv(DRILL_PATH / 'drlrun.csv')
dr_sg      = pd.read_csv(DRILL_PATH / 'sg.csv')
dr_survey  = pd.read_csv(DRILL_PATH / 'survey.csv')

print('Archivos cargados OK')
```

## 2. Resumen de datasets

```python
datasets = {
    'BH collar':    bh_collar,
    'BH assay':     bh_assay,
    'BH litho':     bh_litho,
    'BH survey':    bh_survey,
    'Drill collar': dr_collar,
    'Drill assay':  dr_assay,
    'Drill lith':   dr_lith,
    'Drill drlrun': dr_drlrun,
    'Drill sg':     dr_sg,
    'Drill survey': dr_survey,
}

summary = pd.DataFrame([
    {'Dataset': k, 'Filas': v.shape[0], 'Columnas': v.shape[1],
     'Columnas_lista': ', '.join(v.columns.tolist())}
    for k, v in datasets.items()
])
summary
```

## 3. Nulos por columna — BH Assay (variable principal: CU)

```python
null_pct = bh_assay.isnull().mean().mul(100).round(1).sort_values(ascending=False)
print(null_pct.to_string())
```
